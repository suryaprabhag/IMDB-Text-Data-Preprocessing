# Task 2: Text Cleaning

For this task, creating a new column called cleaned_review rather than modifying the original review column. This is better approach because we can compare before vs. after preprocessing.


---

## 1. Convert text to lowercase

```python
df['cleaned_review'] = df['review'].str.lower()

df[['review', 'cleaned_review']].head()
```

## 2. Remove punctuation marks

```python
df['cleaned_review'] = df['cleaned_review'].str.replace(
    f'[{string.punctuation}]',
    '',
    regex=True
)
```

```python
print(df['cleaned_review'].iloc[0])
```

## 3. Remove special characters

We can use regular expressions:

```python
# This keeps English letters a-z Spaces and removes unwanted special characters.
df['cleaned_review'] = df['cleaned_review'].str.replace(
    r'[^a-zA-Z\s]',
    '',
    regex=True
)
```

## 4. Remove numerical values

```python
df['cleaned_review'] = df['cleaned_review'].str.replace(
    r'\d+',
    '',
    regex=True
)
```

**Note:**

The previous special-character step already removes digits because of [^a-zA-Z\s]. Therefore, this numerical-removal step may not visibly change the text. That's okay—you can still show it as a separate preprocessing requirement in your assignment.

## 5. Remove extra whitespaces

```python
df['cleaned_review'] = df['cleaned_review'].str.replace(
    r'\s+',
    ' ',
    regex=True
).str.strip()
```

# Task 3: Tokenization

`Objective`

Tokenization means splitting a sentence into individual words or tokens.

`Explanation`

Tokenization converts continuous text into individual units called tokens. This makes the text easier to process for subsequent NLP techniques such as stop-word removal, stemming, and lemmatization.

```python
df['tokens'] = df['cleaned_review'].apply(word_tokenize)
```

### Display sample output

```python
df[['cleaned_review', 'tokens']].head()
```

# Task 4: Stop Word Removal

Stop words are commonly occurring words that generally contribute less useful information for many NLP tasks.

`Examples:`

the, is, am, are, was, were, a, an, the, of, in, on, and

## Load English stop words

```python
stop_words = set(stopwords.words('english'))

# Keep important negation words for sentiment analysis
negation_words = {'not', 'no', 'never'}

# Remove negation words from the stop-word list
stop_words = stop_words - negation_words

print("Number of stop words:", len(stop_words))
print(list(stop_words)[:20])
```

## Remove stop words

`Explanation`

Stop-word removal reduces unnecessary/common words and can reduce the number of tokens that a machine-learning model needs to process.

Important: Stop-word removal is task-dependent. Words such as not can be important for sentiment analysis.

`For example:`

Removing not could change the meaning. For a sentiment-analysis project, you can mention this as an observation in your report.

```python
df['tokens_no_stopwords'] = df['tokens'].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)
```

### Display sample output

```python
for i in range(3):
    print("Before Stop Word Removal:")
    print(df['tokens'].iloc[i])

    print("\nAfter Stop Word Removal:")
    print(df['tokens_no_stopwords'].iloc[i])

    print("=" * 100)
```

## Task 5: Stemming

Stemming reduces words to their root or stem by removing prefixes/suffixes.

`For example:`

- playing → play
- played → play
- studies → studi

Notice that the stem doesn't always have to be a proper English word.

`Explanation`

Stemming reduces different forms of a word to a common stem. It is computationally fast, but the resulting stem may not always be a valid English word.

`Create stemmer`

```python
stemmer = PorterStemmer()
```

## Apply stemming

```python
df['stemmed_tokens'] = df['tokens_no_stopwords'].apply(
    lambda tokens: [stemmer.stem(word) for word in tokens]
)
```

```python
for i in range(3):
    print("Before Stemming:")
    print(df['tokens_no_stopwords'].iloc[i])

    print("\nAfter Stemming:")
    print(df['stemmed_tokens'].iloc[i])

    print("=" * 100)
```

## Task 6: Lemmatization

Lemmatization is similar to stemming, but it attempts to convert words to their meaningful dictionary/base form.

`For example:`

- running → running/run
- better → better
- studies → study

`Create lemmatizer`

```python
lemmatizer = WordNetLemmatizer()
```

## Apply lemmatization

```python
df['lemmatized_tokens'] = df['tokens_no_stopwords'].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)
```

```python
for i in range(3):
    print("Before Lemmatization:")
    print(df['tokens_no_stopwords'].iloc[i])

    print("\nAfter Lemmatization:")
    print(df['lemmatized_tokens'].iloc[i])

    print("=" * 100)
```

# Task 7: Final Preprocessed Text

We will use the lemmatized tokens because lemmatization generally produces more meaningful words than stemming.

`Convert tokens back into text`

```python
df['final_preprocessed_text'] = df['lemmatized_tokens'].apply(
    lambda tokens: ' '.join(tokens)
)
```

# Task 8: Summary


**Techniques Applied**

 **1. Lowercasing** – Converted all text to lowercase to maintain consistency and avoid treating uppercase and lowercase versions of the same word as different words.

**2. Punctuation Removal** – Removed punctuation marks such as ., ,, !, ?, etc., because they are generally not required for basic text classification.

**3. Special Character Removal** – Removed unwanted special characters and retained alphabetic characters and spaces.

**4. Numerical Value Removal** – Removed numbers because they were not considered important for the basic sentiment-analysis task.

**5. Extra Whitespace Removal** – Removed unnecessary spaces to make the text cleaner and more consistent.

**6. Tokenization** – Split each review into individual words/tokens so that each word could be processed separately.

**7. Stop Word Removal** – Removed frequently occurring words that may contribute less information to the model. However, for sentiment analysis, words such as "not" can be important and should be handled carefully.

**8.Stemming** – Reduced words to their common stems. It is fast but can sometimes generate non-dictionary words.

**9. Lemmatization** – Converted words into their meaningful base/dictionary forms. It generally produces more linguistically meaningful results than stemming.


**Observations**

After preprocessing, the reviews became shorter and more standardized. Uppercase letters, punctuation, numbers, special characters, and unnecessary spaces were removed. Tokenization separated the reviews into individual words, while stop-word removal reduced common words. Stemming and lemmatization reduced variations of words to common forms.

For this project, lemmatized text is selected as the final preprocessed text because it generally retains more meaningful linguistic information than stemming.

**Observation:**

 Stop-word removal was performed while retaining important negation words such as "not", "no", and "never", as these words can significantly affect sentiment. This prevents the preprocessing step from unintentionally changing the meaning of a review.
